```python
class BankAccount:
    def __init__(self):
        self.balance = 0

    def deposit(self, amount):
        self.balance += amount
        print("Balans:", self.balance)

    def withdraw(self, amount):
        self.balance -= amount
        print("Balans:", self.balance)

account = BankAccount()
account.deposit(1000)
account.withdraw(300)
```

```python
class Studentgrades:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def show(self):
        print("Talaba:", self.name)
        print("Eng yuqori:", max(self.grades))
        print("Eng past:", min(self.grades))
        print("Ortacha:", sum(self.grades) / len(self.grades))

s = Studentgrades("Ali", [80, 90, 70])
s.show()
```

```python
class Myprofile:
    def __init__(self):
        self.name = "Yusuf"
        self.age = 18

    def show(self):
        print("Ism:", self.name)
        print("Yosh:", self.age)

me = Myprofile()
me.show()
```
